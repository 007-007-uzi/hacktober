```python
import torch
from transformers import AutoTokenizer, AutoModelForSequenceClassification
model_id = "distilbert/distilbert-base-uncased-finetuned-sst-2-english"
tokenizer = AutoTokenizer.from_pretrained(model_id)
model = AutoModelForSequenceClassification.from_pretrained(model_id)
model.eval()
print("MoodLens AI model is ready!")
```

```python

def analyze_sentiment(text):
    # Step 1: Convert text into numbers
    inputs = tokenizer(
        text,
        return_tensors="pt",
        truncation=True
    )
```

```python
def analyze_sentiment(text):
    # Step 1: Convert text into numbers
    inputs = tokenizer(
        text,
        return_tensors="pt",
        truncation=True
    )

    # Step 2: Model Inference
    with torch.no_grad():
        outputs = model(**inputs)

    # Step 3: Find the predicted category
    probabilities = torch.softmax(
        outputs.logits, dim=-1
    )

    predicted_id = probabilities.argmax(dim=-1).item()
    label = model.config.id2label[predicted_id]
    confidence = probabilities[0][predicted_id].item()

    return label, confidence

print(analyze_sentiment("I love learning AI!"))
```

```python
print("Welcome to MoodLens!")
print("Type 'exit' to stop.\n")

while True:
    text = input("Enter an English sentence: ").strip()

    if text.lower() == "exit":
        print("Thanks for trying MoodLens!")
        break

    if not text:
        print("Please enter a sentence.")
        continue

    label, confidence = analyze_sentiment(text)

    print("\nAI Prediction:", label)
    print(f"Model confidence: {confidence:.1%}")
    print("-" * 35)
```
